3. 產生的圖片會自動儲存到您的 Google 雲端硬碟 `2026AI_Agent/教學駕駛艙/units/g5-cyber-safety/assets/`，電腦本機會自動同步！

```python
# 第一步：安裝所需套件與掛載 Google Drive
!pip install -q diffusers transformers accelerate torch torchvision

import os
import torch
from google.colab import drive
from diffusers import AutoPipelineForText2Image

drive.mount('/content/drive')

# 建立本專案在 Google Drive 的存放路徑
save_dir = '/content/drive/MyDrive/2026AI_Agent/教學駕駛艙/units/g5-cyber-safety/assets'
os.makedirs(save_dir, exist_ok=True)
print(f'檔案將會自動儲存至：{save_dir}')
```

```python
# 第二步：載入 SDXL-Turbo 模型（約需 30 秒至 1 分鐘下載權重）
print('正在載入模型至 GPU...')
pipe = AutoPipelineForText2Image.from_pretrained(
    'stabilityai/sdxl-turbo',
    torch_dtype=torch.float16,
    variant='fp16'
).to('cuda')
print('模型載入完成！')
```

```python
# 第三步：設定繪製任務清單（依情境自然搭配男女學生與長輩）
tasks = [
    {
        'case': 'case4',
        'name': '案件 4 可疑連結（小女孩冷靜辨識假連結）',
        'filename': 'case4_phish_girl.png',
        'prompt': 'Cute 3D illustration style, Pixar and storybook aesthetic, soft warm studio lighting, clean solid pure white background (#FFFFFF). A clever Taiwanese elementary school girl with neat black hair in ponytail, wearing casual modern clothes, holding smartphone and cautiously inspecting a floating hologram alert bubble of a suspicious web link with a magnifying glass, cybersecurity fact-checking concept, isolated on white background, sharp focus, 8k'
    },
    {
        'case': 'case5',
        'name': '案件 5 阿嬤接到假電話（小男孩與阿嬤約定通關密語）',
        'filename': 'case5_grandma_passphrase.png',
        'prompt': 'Cute 3D illustration style, Pixar aesthetic, soft warm studio lighting, clean solid pure white background (#FFFFFF). A kind smiling Taiwanese grandmother with silver hair holding a home telephone handset, standing happily beside her polite 10-year-old grandson with neat black hair holding a small notebook with secret family passphrase, warm family cybersecurity concept, isolated on white background, sharp focus, 8k'
    },
    {
        'case': 'case7',
        'name': '案件 7 搞笑貼圖風波（男女同學討論肖像權與同理心）',
        'filename': 'case7_photo_dilemma.png',
        'prompt': 'Cute 3D illustration style, Pixar aesthetic, soft warm studio lighting, clean solid pure white background (#FFFFFF). Two Taiwanese elementary school students, a boy with neat black hair in green sweater holding a tablet, and a girl with black hair looking thoughtful, discussing portrait privacy and empathy before sharing funny meme photos, school cybersecurity ethics concept, isolated on white background, sharp focus, 8k'
    },
    {
        'case': 'case10',
        'name': '案件 10 遊戲代儲陷阱（小男孩機警關閉免費點數彈窗）',
        'filename': 'case10_game_scam.png',
        'prompt': 'Cute 3D illustration style, Pixar aesthetic, soft warm studio lighting, clean solid pure white background (#FFFFFF). A clever Taiwanese elementary school boy with neat black hair sitting at desk in front of a computer monitor showing a game discount pop-up, smartly moving mouse pointer to click the red close X button, rejecting scam traps, isolated on white background, sharp focus, 8k'
    },
    {
        'case': 'case11',
        'name': '案件 11 網路霸凌與溫暖求助（小女孩關心陪伴同學）',
        'filename': 'case11_anti_bully.png',
        'prompt': 'Cute 3D illustration style, Pixar aesthetic, soft warm studio lighting, clean solid pure white background (#FFFFFF). A kind Taiwanese elementary school girl warmly comforting a sad classmate sitting on a bench, handing a tissue with empathy and friendship, positive support against cyberbullying, isolated on white background, sharp focus, 8k'
    }
]

# 開始批次生成並直接存入 Google Drive
from IPython.display import display

for t in tasks:
    print(f'\n🎨 正在生成：{t["name"]}...')
    # SDXL-Turbo 只要 2~4 步，guidance_scale=0.0，速度極快且畫質極高！
    img = pipe(prompt=t['prompt'], num_inference_steps=4, guidance_scale=0.0).images[0]
    out_path = os.path.join(save_dir, t['filename'])
    img.save(out_path)
    print(f'✅ 已成功儲存至 Google Drive：{t["filename"]}')
    display(img)

print('\n🎉 所有案件插畫生成完畢！您的本機電腦會自動同步這些檔案！')
```
